```python
def mse(face1,face2):
    squared_error_sum=0
    for i in range(len(face1.embedding)):
        k=abs(face1.embedding[i]-face2.embedding[i])
        k=k**2
        squared_error_sum += k

    n=len(face1.embedding)
    return squared_error_sum/n
```

```python
import numpy as np
def cosine_similarity(face1,face2):
    a=face1.embedding
    b=face2.embedding
    return np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
```

```python
def find(image_path):
    img=cv2.imread(image_path)
    faces=app.get(img)
    if len(faces) != 1:
        return "Image must contain exactly one face"
    face=faces[0]
    embeddings=face.embedding
    result=""
    max_k=0
    for faceid,embedding in data.items():
        k=cosine_similarity(embeddings,embedding)
        if(k>0.50):
            if k>max_k:
                result=faceid
                max_k=k

    if result=="":
        return "NO result found"

    return result
    
    
```
